**FUNCIÓN evaluar_audiencia PARA AMBOS ALGORITMOS**

```
def evaluar_audiencia(poblacion):
  evaluaciones = []
  for individuo in poblacion:
    audiencia_total = 0
    audiencia_partidos = []
    repetidos_horarios = {} #contamos cuantas veces aparece cada horario, en un diccionario

    for partido, horario in individuo:
      equipo1, equipo2 = partido #separamos la tupla formada por cada equipo, pasando cada uno a una variable para buscar su categoria
      categoria_equipo_1 = equipos[equipo1]
      categoria_equipo_2 = equipos[equipo2]
      if(categoria_equipo_1) > (categoria_equipo_2):
        categoria_por_partido = (categoria_equipo_2 +"-"+ categoria_equipo_1)
      else:
        categoria_por_partido = (categoria_equipo_1 +"-"+ categoria_equipo_2)

      repetidos_horarios[horario] = repetidos_horarios.get(horario, 0) + 1
      audiencia_sin_correcion = audiencias_mill[categoria_por_partido]
      coeficiente_horario = c_horarios[horario]
      audiencia_correcion_horario = audiencia_sin_correcion * coeficiente_horario
      audiencia_partidos.append({
      "partido": partido,
      "horario": horario,
      "categoria": categoria_por_partido,
      "audiencia": audiencia_correcion_horario
      })

    for valor in audiencia_partidos:
      horario = valor["horario"]
      audiencia = valor["audiencia"]
      c_repetido = c_coincidencia[repetidos_horarios[horario]]
      audiencia_final = audiencia * c_repetido
      valor["repeticiones"] = repetidos_horarios[horario]
      valor["audiencia_final"] = audiencia_final
      audiencia_total += audiencia_final

    evaluaciones.append((individuo, audiencia_total))
    #Este sería nuestro fitness, la audiencia total es el fitness de cada individuo de la población

  return evaluaciones
```

**SOLUCIÓN**

Para resolver el algoritmo por fuerza bruta se va usar *https://docs.python.org/es/3/library/itertools.html*, en concreto la función *product* y cuya documentación específica: "producto cartesiano, equivalente a un bucle for anidado", ideal para resolver este algoritmo por fuerza bruta y evitar poner 10 bucles anidados, que sería la otra opción.

```
from itertools import product
max_audiencia = 0
mejor_individuo = []
contador = 0 #ponemos un contador porque viendo las combinaciones que nos han salido al caluclarlas, es inviable que se ejecute completamente
max_repes = 50000 #ponemos 50000 repeticiones como máximo para que pare en algún momento
for x in product(c_horarios.keys(), repeat = 10):
  contador += 1
  if contador > max_repes:  #Por si da algún resultado invalido, sin L o V
    break
  if "V20" not in x or "L20" not in x:
    continue
  #Como la función evaluar_audiencia se concibio para el algoritmo de optimización y posteriormente se desarrollo este, le pasaba una población, en este caso para no cambiarle, directamente metemos
  #cada individuo en una lista, y asi sería una población de solo un individuo y nos permite compararlo
  individuo = list(zip(jornada_prefijada, x)) #juntamos en una lista los partidos, junto con el horario
  evaluacion = evaluar_audiencia([individuo]) #le pasamos al individuo dentro de otra lista, como hemos comentado
  audiencia = evaluacion[0][1] #cogemos del primer elemento de la lista, en este caso el único, el valor de audiencia que tiene
  if audiencia > max_audiencia:
    max_audiencia = audiencia
    mejor_individuo = individuo

print(f"Mejor individuo: {mejor_individuo} con un máximo de audiencia de {max_audiencia}")
```

Para la solución de este apartado, vamos a usar un algoritmo genético, en el que hemos decicido las siguientes cuestiones:

•individiuo: es una jornada de liga completa, no es lo mismo que un partido.


•población: serían varias propuestas de la misma jornada, pero en referente al horario, no en cuanto a los enfrentamientos que son los mismos para todos los individuos de la población.


•gen: sería un partido con un horario asignado.


El algoritmo básicamente lo que hace basado en una población elegida de 200 individuos, primero se genera esa población. Posteriormente cogemos un grupo de 3 individuos, de los cuales cogemos el mejor y lo consideramos padre, y mismo proceso para la madre, eso sí evitando que pudieran ser el mismo individuo.

Una vez que hemos cruzado a ambos padres, siempre dan dos hijos, estos hijos tienen una parte del padre de entre el 30% y el 70%, y el resto de la madre. Cada hijo tiene una probabilidad de mutar, del 10%, y esta mutación se produce en solamente un gen, es decir en un partido asignado.Cuando sabemos si el hijo ha mutado o no, se evalua que cumpla los requisitos de que se juegue un partido tanto el Viernes como el Lunes, sino es así se asigna un partido aleatorio a esos días, y si de por sí el hijo ya tenía mas de uno esos días, se mantiene (es posible que todos ellos tengan varios partidos en la misma franja horaria). La mutación se agrega para evitar posibles estancamientos en máximos locales, ya que introduce mayor diversidad en la población.

Una vez que hemos generado hijos, vamos rellenando la población de 200 individuos definida. El mejor de cada generación pasa si o si a la siguiente, aunque sea padre, es decir se usa elitismo. Si por ejemplo como se generan 2 hijos en una población de 200, y ya tenemos un individuo de la población anterior(el mejor) lo que se hace para rellenar el último hueco en 199 y que sobraría un hijo, se generan ambos hijos y se coge el mejor de ellos.

Como criterio de parada hemos establecido dos, es decir probar 1000 generaciones aunque se han hecho pruebas con muchas diferentes, y como criterio extra es que haya multiples generaciones sin mejorar, en este caso fijada como 500, en cuyo caso si tras 500 generaciones y no habiendo llegado a 1000 no se ha mejorado el algoritmo abandona. Este criterio de parada es el que realmente sucede.

**FUNCIONES PROPIAS DEL ALGORITMO GENÉTICO**

```
def comprobar_hijos_validos(hijo):
  numero_viernes = 0
  numero_lunes = 0
  indice_viernes = None
  indice_lunes = None
  for i, (partido, horario) in enumerate(hijo):
    if horario == "V20":
      numero_viernes +=1
      indice_viernes = i
    elif horario == "L20":
      numero_lunes +=1
      indice_lunes = i

  if numero_viernes == 0:#Miramos el numero de viernes
    indice_nuevo_viernes = random.randrange(len(hijo)) #cogemos un indice aleatorio para poner el partido de viernes porque no hay ninguno.
    if numero_lunes == 1:  #revisamos, si el número de lunes es 1 para no tocarlo, si hay mas da igual sustituirle, y si hay 0 luego corregimos el lunes.
      while indice_nuevo_viernes == indice_lunes: #mientras que sea igual el indice, seguimos generando otro indice.
        indice_nuevo_viernes = random.randrange(len(hijo))

    partido = hijo[indice_nuevo_viernes][0]
    hijo[indice_nuevo_viernes] = (partido, "V20") #guardamos el nuevo horario
    indice_viernes = indice_nuevo_viernes #guardamos en indice_viernes el que acabamos de meter, por si acaso no había ninguno tener este ya dentro
    numero_viernes +=1

  if numero_lunes == 0:
    indice_nuevo_lunes = random.randrange(len(hijo))
    if numero_viernes == 1:
      while indice_nuevo_lunes == indice_viernes:
        indice_nuevo_lunes = random.randrange(len(hijo))

    partido = hijo[indice_nuevo_lunes][0]
    hijo[indice_nuevo_lunes] = (partido, "L20")
    #numero_lunes += 1
    #indice_lunes = indice_nuevo_lunes
  return hijo

def cruzar_descendencia(individuo1, individuo2):
  corte = random.randint(3,7) #Para no coger siempre la mitad de cada individuo, generamos un numero aleatorio entre 3-7, así como mínimo cogera 3 genes y como máximo 7
  hijo1 = individuo1[:corte] + individuo2[corte:]
  hijo2 = individuo2[:corte] + individuo1[corte:]
  hijo1 = comprobar_hijos_validos(hijo1)
  hijo2 = comprobar_hijos_validos(hijo2)
  return hijo1, hijo2

def mutar(hijo, horarios):  #En la mutación cambiamos solamente un partido
  indice = random.randrange(len(hijo)) #generamos un número aleatorio
  partido = hijo[indice][0] #cogemos el partido que pertenece a ese indice
  horario_actual = hijo[indice][1] #cogemos el horario de ese partido
  horas_posibles = [ horario for horario in horarios if horario != horario_actual] #a traves de list comprehension elegimos un horario que sea distinto al actual para garantizar que mute
  nuevo_horario = random.choice(horas_posibles) #elegimos una opción aleatoria entre las horas posbiles
  hijo[indice] = (partido, nuevo_horario)
  hijo = comprobar_hijos_validos(hijo)
  return hijo

def seleccion_torneo(poblacion_anterior, tamaño):
  posibles = random.sample(poblacion_anterior,tamaño)
  ganador = max(posibles, key = lambda x: x[1])
  return ganador


def crear_individuo_aleatorio(jornada, horarios):
  partidos = []
  partido_viernes, partido_lunes = random.sample(range(len(jornada)),2) #coge 2 partidos al azar DISTINTOS, que serán el del lunes y de viernes
  for i, partido in enumerate(jornada):
    #por cada elemento en jornada que hemos generado, comprobamos si es el indice aleatorio que nos ha tocado para lunes/viernes se le asigna ese horario, sino se coge una opcion random y guardamos en la lista partidos.
    if i == partido_lunes:
      hora = "L20"
    elif i == partido_viernes:
      hora = "V20"
    else:
      hora = random.choice(horarios)

    partidos.append((partido, hora))

  return partidos
```

**ALGORITMO GENÉTICO**

```
def algoritmo_genetico(mejor, poblacion_anterior):
  nueva_poblacion = [mejor[0]] # mejor es individuo, puntuacion, asi que guardamos solo el individuo, la puntuación como tenemos que calcularla para el resto de población
  #lo primero de todo traemos al mejor de la poblacion generada aleatoriamente, como parte de usar el criterio de ELITISMO
  #Comenzaremos con lo definido com otorneo, donde se elegira poblacion al azar y de estos se seleccionara el mejor para repetir el mismo proceso para hallarle una pareja, de estos saldra un descendiente.
  #La idea es quedarnos ocn el mejor de la poblacion anterior, y la descendencia de la poblacion_anterior hasta hacer otra poblacion nueva de 200 indidiuos.
  while len(nueva_poblacion) < len(poblacion_anterior):
    mejor_seleccion_padre = seleccion_torneo(poblacion_anterior, 3) #llamamos a la función de selección por torneo, elegimos 3 al azar y nos quedamos con el mejor
    mejor_seleccion_madre = seleccion_torneo(poblacion_anterior, 3)
    while mejor_seleccion_madre[0] == mejor_seleccion_padre[0]: #while por si llegamos a elegir a los dos iguales, entonces volvemos a seleccionar uno diferente para la madre, mientras no lo encuentre no para.
      mejor_seleccion_madre = seleccion_torneo(poblacion_anterior, 3)
    hijo1, hijo2 = cruzar_descendencia(mejor_seleccion_padre[0], mejor_seleccion_madre[0])
    #Aquí introducimos las mutaciones, porque hemos comprobado que en 50 rondas o generaciones se queda atascada a partir de la 10-11 aproximadamente, de esta manera le damos un componente pequeño del 10% de aleatoriedad
    if random.random() < 0.1:
      hijo1 = mutar(hijo1, horarios)
    if random.random() < 0.1:
      hijo2 = mutar(hijo2, horarios)

    #Como estamos siendo elitistas, y guardamos el mejor de la poblacion anterior, y si queremos respestar el tamaño de la población, va a llegar un punto en el que
    #va a quedar unhijo fuera, entonces controlamos este punto, evaluamos los hijos, comparamos cual es mejor de los 2 últimos hijos, y únicamente guardamos ese.
    if len(nueva_poblacion) == len(poblacion_anterior) - 1:
      poblacion_hijos = [hijo1, hijo2]
      evaluacion_hijos = evaluar_audiencia(poblacion_hijos)
      mejor_hijo = max(evaluacion_hijos, key=lambda x: x[1])
      nueva_poblacion.append(mejor_hijo[0])
    else:
      nueva_poblacion.append(hijo1)
      nueva_poblacion.append(hijo2)

  return nueva_poblacion

#/////////////////////# EJECUCIÓN DEL ALGORITMO GENÉTICO#/////////////////////#
poblacion = []
horarios = list(c_horarios.keys())
#creamos una muestra de 200 individuos aleatorios completamente, aunque más adelante pida que genere entradas aleatorias, entendemos que esto es básico de cualquier algoritmo genético, y se hace a través
#de crear_individuo_aleatorio y recibe la jornada que es la que generaremos aleatoramiente en el apartado de más adelante, y el horario que viene marcado por el enunciado.
for x in range(200):
  individuo = crear_individuo_aleatorio(jornada_prefijada, horarios)
  poblacion.append(individuo)

evaluacion = evaluar_audiencia(poblacion)
mejor = max(evaluacion, key = lambda x: x[1])

#print(mejor)
#algoritmo_genetico(mejor, evaluacion)
max_generaciones = 1000
limite_sin_mejora = 500
num_generaciones_sin_mejora = 0
mejor_generacion = 0

poblacion_actual = evaluacion #la población actual es la que hemos recibido la generación aleatoria, elegimos al mejor de esta población, y también marcamos un marcador para guardar el mejor final.
mejor_poblacion_actual = mejor #mejor es el valor que traiamos del mejor de la generacion generada random
mejor_final = mejor

for generacion in range(max_generaciones):
  nueva_poblacion = algoritmo_genetico(mejor_poblacion_actual, poblacion_actual)
  poblacion_actual = evaluar_audiencia(nueva_poblacion)
  mejor_poblacion_actual = max(poblacion_actual, key=lambda x: x[1])

  if mejor_poblacion_actual[1] > mejor_final[1]:
    mejor_final = mejor_poblacion_actual
    num_generaciones_sin_mejora = 0
    mejor_generacion = generacion +1 #guardamos como mejor generacion la actual, que corresponde con el lugar que ocupa el indice del for +1
  else:
    num_generaciones_sin_mejora += 1

  print(f"Generación {generacion +1} -> con la audiencia de {mejor_poblacion_actual[1]:.3f}")

  if num_generaciones_sin_mejora >= limite_sin_mejora:
    print(f"No hay mejora en {limite_sin_mejora} generaciones, por lo que se para de ejecutar")
    break

print(f"La mejor audiencia es: {mejor_final[1]}, en la generación {mejor_generacion}")
print(f"La mejor distribución es:")
for partido, horario in mejor_final[0]:
  print(partido, "--", horario)
```

Dentro de cada generación se procesan P individuos(la población). Para cada uno de los individuos además se procesan los N partidos para poder calcular la audiencia. Por tanto, la complejidad de los algoritmos genéticos depende del número de generaciones, del tamaño de la población P y del número de partidos N, $O(G \cdot P \cdot N)$.
